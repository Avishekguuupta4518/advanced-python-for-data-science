```python
import pandas as pd 
df_A = pd.DataFrame({
    "Temperature" : [20, 30, 40, 50, 60]}, 
    index=['Day1', 'day2' , 'day3', 'day4', 'day5']
)

df_B = pd.DataFrame({
    "Humidity" : [120, 130, 140, 150, 160],}, 
    index=['Day1', 'day2' , 'day3', 'day5', 'day7']
)

joined_df = df_A.join(df_B, how='inner')
print(joined_df)
```

```python
joined_df = df_A.join(df_B, how='outer')
print(joined_df)
```

```python
joined_df = df_A.join(df_B, how='left')
print(joined_df)
```

```python
joined_df = df_A.join(df_B, how='right')
print(joined_df)
```

```python
wide_df = pd.DataFrame({
    "Name": ["Avishek", "Rahul"],
    "Science" : [20,23],
    "English" : [45,64],
    "Math" : [63,24]
})

print(wide_df)
```

```python
long_df = pd.melt(
    frame=wide_df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'   
)
print(long_df)
```

```python
new_df = long_df.pivot(
    index='Name',
    columns='Subject',
    values="Marks"
)

print(new_df)
```

```python
stacked_df = long_df.stack()
print(stacked_df)
```

```python
unstacked_df = stacked_df.unstack()
print(unstacked_df)
```

```python
pivot_table_df = pd.pivot_table(
    data=long_df,
    index='Name',
    columns='Subject',
    values="Marks",
    aggfunc='mean'
)

print(pivot_table_df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "EID": [101,102,103,104],
    "Name": ["Ram", "Shyam", "Sita", "Krishna"],
    "Level": ["Junior", "Senior", "Mid", "Junior"]
})


df.info()
```

```python
df.dtypes
```

```python
df['Level'] = df['Level'].astype('category')
df.dtypes
```

```python
df['Level'] = df['Level'].cat.rename_categories({
    'Junior':'Jr',
    'Senior':'Sr'
})

df.head()
```

```python
df['Level_code'] = df['Level'].cat.codes
print(df.head())
```

```python
encoded = pd.get_dummies(df['Level'], prefix='Level')
df = pd.concat(objs= [df, encoded],axis=1)
print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Date':['2025-01-11', '2026-01-12', '2005-12-05'],
    'Sales':[101,102,103]
})

print(df.dtypes)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
print(df.dtypes)
```

```python
df = df.set_index('Date')
print(df.head())
```

```python
df['Year'] = df.index.year
df['Month'] = df.index.month
df['Day'] = df.index.day
df['Day_Name'] = df.index.day_name()
df['Month_Name'] = df.index.month_name()
print(df.head())
```

```python
df.loc['2005']
```

```python
df.plot.bar(x='Day_Name',y='Sales')
```
