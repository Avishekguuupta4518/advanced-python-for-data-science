```python
import pandas as pd

df_A = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Avishek", "Rahul", "Sita", "Anita"]
})

df_B = pd.DataFrame({
    "Roll": [3, 4, 5, 6],
    "Age": [20, 21, 22, 23]
})

joined_df = df_A.join(df_B, how='inner')
print(joined_df)
```
