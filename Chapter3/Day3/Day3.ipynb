### Normalization

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

X = np.array([[10], [20], [30], [40], [100]])
scaler = MinMaxScaler(feature_range=(0,1))
X_transformed = scaler.fit_transform(X)

print(X_transformed)
```

### Z-score Normalization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X = np.array([[10], [20], [30], [40], [100]])

X_transformed = scaler.fit_transform(X)

print(X_transformed)
```

### Decimal Scaling Normalization

```python
X = np.array([[10], [20], [30], [40], [-100]])

j = len(str(np.max(np.abs(X))))
X_transformed = X/10**j
print(X_transformed)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
cities = ['Kathmandu', 'Pokhara', 'Biratnagar']
encoded = le.fit_transform(cities)

print(encoded)
```

## Feature Encoding One hot encoding

```python
import pandas as pd

df = pd.DataFrame({'source': ['river', 'lake', 'well']})
df_encoded = pd.get_dummies(df, columns=['source'])

print(df_encoded)
```

## Ordinal encoding

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['poor', 'fair', 'good']])
quality = [['poor'], ['good'], ['fair']]
encoded = encoder.fit_transform(quality)

print(encoded)
```

```python
import numpy as np
import time

X = np.random.random(10000000)
print(X[:5])
start = time.time()
for i in range(X.shape[0]):
    X[i] = X[i] + 2
end = time.time()    
print(X[:5]) 
print(f'Time taken is {end-start} second')   
```
