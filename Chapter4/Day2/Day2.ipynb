### Chi-Square Test

```python
import numpy as np

#observed counts
observed = np.array([
    [250, 200],
    [50, 1000]
])

# Expected counts
expected = np.array([
    [90, 360],
    [210, 840]
])

# Manually calculate chi- square
chi_square = ((observed - expected)**2 / expected).sum()
print(f"Chi-square value :  {chi_square:.4f}")
```

```python
from scipy.stats import chi2_contingency

chi2, p, dof, exp = chi2_contingency(observed, correction=False)

print(f"Chi-Square : {chi2:.4f}")
print(f"p-value: {p:.4f}")
print(f"Degree of freedom: {dof}")
print(f"Expected Value: {exp}")

if p < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
from scipy.stats import chisquare

observed = [16, 20, 25, 14, 29, 28]
expected = [22] * 6

chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print(f"Chi-Square : {chi2_stat:.4f}")
print(f"p-value: {p_value:.4f}")

if p < 0.05:
    print("Reject H0. The die is biased")
else:
    print("Fail to reject H0. The die is unbiased")
```

### Regression

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# Generate values from 0 to 30 with
X = np.arange(0,30,0.27)
print(X[0:4])

# generate y = mx+c
y = 0.1999*X + 4.2999
# view the first 4 value from y
print(y[0:4])
# View the shape of X and y
print(X.shape, y.shape)
```

```python
split = int(X.shape[0]*0.8)
print(split)
```

```python
X_train, y_train = X[0:split], y[0:split]
X_test, y_test = X[split:], y[split:]

```

```python
def plot_data(X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test, predicitions=None):
    plt.scatter(X_train, y_train, c='b', label='Traning Data')
    plt.scatter(X_test, y_test, c='r', label='Testing Data')
    
    if predicitions is not None:
        plt.plot(X_test, predicitions, c='g', label='Predicitions')
    plt.legend()
    plt.title("X vs y")
    plt.show()
    
plot_data()
```

```python
X_mean = np.mean(X_train)
y_mean = np.mean(y_train)
cov_x_y = (1/X_train.shape[0]) * np.sum((X_train-X_mean)*(y_train-y_mean))
cov_x = (1/X_train.shape[0]) * np.sum(np.square(X_train-X_mean))

m = cov_x_y/cov_x
c=y_mean- m*X_mean
print(f'y = {m}*X+{c}')
```

```python
y_preds = m*X_test +c 
plot_data(predicitions=y_preds)
```

```python
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()

# Train 
lr_model.fit(X_train.reshape(-1,1), y_train)

# get predicitions 
y_preds = lr_model.predict(X_test.reshape(-1, 1))

# Plot
plot_data(predicitions=y_preds)
```

### multiple linear regression

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from sklearn.model_selection import train_test_split

X1 = np.arange(10, 30, 0.27)
X2 = np.arange(50, 70, 0.27)
y = 1.12*X1 + 2.123*X2 + 3.123

df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'y': y
})

X_data = df.drop(columns=['y'])
y_data = df['y']

X_train, X_test, y_train, y_test = train_test_split(
    X_data, y_data, test_size=0.2, random_state=42
)

X_train_sm = sm.add_constant(X_train)
X_test_sm = sm.add_constant(X_test)

# fit the model
model = sm.OLS(y_train, X_train_sm).fit()

# view the summary
print(model.summary())

# prediction
y_preds = model.predict(X_test_sm)

```

```python
from sklearn.metrics import r2_score
y_preds=model.predict(X_test_sm)
print(r2_score(y_test, y_preds))
```
