```python
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Fare"].fillna(df["Fare"].median(), inplace=True)
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
df.drop(columns=["Cabin"], inplace=True)
```

```python
df
```

```python
import matplotlib.pyplot as plt

plt.bar(
    x=df['Age'].unique(), 
    height=df['Age'].value_counts(), 
    color=["red", "green", "blue", "yellow"] )
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

plt.figure(figsize=(8,4))
sns.histplot(df['Age'], kde=True, bins=20)
plt.title('Total Age Distribution')
plt.show()
```

```python
sns.boxplot(x='Age', y='Fare', data=df)
plt.title('Total Fare By Age')
plt.show()
```

```python
sns.boxplot(x='Gender', y='Pclass', data=df)
plt.title('Pclass By Gender')
plt.show()
```

```python
sns.scatterplot(
    x='Pclass',
    y='Fare',
    hue='Gender',
    data=df
)

plt.title('Fare vs Pclass by Gender')
plt.show()
```

```python
corr_mat = df.select_dtypes(include="number").corr()

sns.heatmap(
    corr_mat,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    df.select_dtypes(include="number").join(df["Gender"]),
    hue="Gender"
)

plt.show()
```
