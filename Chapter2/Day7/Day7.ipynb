```python
import requests
import pandas as pd
from pandas import json_normalize
from pprint import pprint

city_names = {'Itahari', 'Biratnagar', 'Dharan', 'Janakpur', 'Kathmandu'}
api_key = "31454b4282bc69567483ad1dd8d07af2"

final_list = []

for city in city_names:
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        pprint(data)

        df = json_normalize(data)
        df_weather = df[[
            'name',
            'main.temp',
            'main.humidity',
            'wind.speed',
            'weather'
        ]]

        df_weather['weather_description'] = df_weather['weather'].apply(
            lambda x: x[0]['description']
        )
        df_weather.drop(columns='weather', inplace=True)

        final_list.append(df_weather)

    else:
        print(f"{city} failed | Status Code:", response.status_code)

# Merge all city data
final_df = pd.concat(final_list, ignore_index=True)

# Export to CSV
final_df.to_csv('weather.csv', index=False)

print("✅ Weather data successfully saved to weather.csv")
```
