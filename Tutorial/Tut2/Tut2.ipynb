```python
aList = [1,2,3,4,5]

it = iter(aList)
while True:
    try:
        num = next(it)
        print(num, end=' ')
    except StopIteration:
        break
```

```python
# Iterator for first n even numbers
class CustomCounter:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current < self.limit:
            self.current += 1
            return self.current
        else:
            raise StopIteration
        
counter = CustomCounter(5)

for num in counter:
    print(num, end=' ')
```

```python
def cubes(n):
    for i in range(n):
        yield i**3
        
for val in cubes(5):
    print(val, end=' ')
```

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} functions")
        result = func(*args, **kwargs)
        print(f"Result: {result}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

add(2,3)
```

```python
class EvenNumbers:
    def __init__(self, n):
        self.n = n          
        self.current = 0   
        self.count = 0      

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.n:
            even_number = self.current
            self.current += 2  
            self.count += 1
            return even_number
        else:
            raise StopIteration
        
        
evens = EvenNumbers(5)   
for num in evens:
    print(num)
```

```python
#   Write a generator that yields the Fibonacci sequence up to n terms.

def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

list(fibonacci(10))
```

```python
# Start-End decorator
def start_end_decorator(func):
    def wrapper(*args, **kwargs):
        print("Start")
        result = func(*args, **kwargs)
        print("End")
        return result
    return wrapper


@start_end_decorator
def greet(name):
    print(f"Hello, {name}!")

greet("Avishek")
```

```python
# CSV line-by-line generator

import csv

file_path = "try.csv"

def read_csv(file_path):
    with open(file_path, newline='') as csvfile:
        reader = csv.reader(csvfile)
        for row in reader:
            yield row

for row in read_csv(file_path):
    print(row)
```

```python
# Reverse range iterator

class ReverseRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
        self.current = end - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.start:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

for num in ReverseRange(1, 6):
    print(num)  # 5, 4, 3, 2, 1
```

```python
# Memory usage decorator

import sys

def memory_usage(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        print(f"Memory used by result: {sys.getsizeof(result)} bytes")
        return result
    return wrapper

@memory_usage
def create_list(n):
    return [i for i in range(n)]

create_list(1000)
```

```python
# Infinite prime number generator

def primes():
    D = {}
    q = 2
    while True:
        if q not in D:
            yield q
            D[q*q] = [q]
        else:
            for p in D[q]:
                D.setdefault(p+q, []).append(p)
            del D[q]
        q += 1
```

```python
import itertools
print(list(itertools.islice(primes(), 10)))  # first 10 primes
```

```python
# Memoization decorator

def memoize(func):
    cache = {}
    def wrapper(*args):
        if args in cache:
            return cache[args]
        result = func(*args)
        cache[args] = result
        return result
    return wrapper

@memoize
def factorial(n):
    return 1 if n == 0 else n * factorial(n-1)

print(factorial(5))  # 120
```

```python
# Generator to filter even numbers

def filter_even(lst):
    for num in lst:
        if num % 2 == 0:
            yield num

print(list(filter_even([1,2,3,4,5,6])))  # [2, 4, 6]
```

```python
# Execution logger decorator

import time

def log_execution(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        print(f"Input args: {args}, {kwargs}")
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution time: {end - start:.6f}s")
        return result
    return wrapper

@log_execution
def multiply(a, b):
    return a * b

multiply(5, 6)
```

```python
# Unique items iterator

class UniqueIterator:
    def __init__(self, lst):
        self.lst = lst
        self.seen = set()
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.lst):
            item = self.lst[self.index]
            self.index += 1
            if item not in self.seen:
                self.seen.add(item)
                return item
        raise StopIteration

print(list(UniqueIterator([1,2,2,3,4,3])))  # [1, 2, 3, 4]
```

```python
# Positive input checker decorator

def positive_inputs(func):
    def wrapper(*args):
        if not all(x > 0 for x in args):
            raise ValueError("All inputs must be positive numbers")
        return func(*args)
    return wrapper

@positive_inputs
def add(a, b):
    return a + b

print(add(5, 10))  # works
# add(-5, 10)  # raises ValueError
```

```python
# Infinite random number generator

def infinite_random():
    import random
    while True:
        yield random.random()  # 0.0 <= x < 1.0
```
