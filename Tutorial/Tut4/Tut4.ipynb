```python
# Read a CSV file of sales data, display the first 10 rows, and summarize sales by month.

import pandas as pd

df = pd.read_csv("sales_data.csv")

df['date'] = pd.to_datetime(df['date'])

print("First 10 Rows:")
print(df.head(10))

df['month'] = df['date'].dt.to_period('M')
monthly_sales = df.groupby('month')['sales'].sum()

print("\nMonthly Sales Summary:")
print(monthly_sales)
```

```python
# Handle missing values in a dataset by filling with median values and verify the result.

import pandas as pd

df = pd.read_csv("missing_value_dataset.csv")
```

```python
print("Missing values BEFORE cleaning:\n")
print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("\nMissing values AFTER cleaning:\n")
print(df.isnull().sum())
```

```python
print("\nCleaned Dataset:\n")
print(df)
```

```python
# Implement error handling while reading from an API and log errors.


import requests
import logging
import pandas as pd
from pprint import pprint

logging.basicConfig(
    filename="api_errors.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def fetch_weather_data(latitude, longitude):
    """
    Fetch current weather data from Open-Meteo API.
    
    Args:
        latitude (float): Latitude of the location
        longitude (float): Longitude of the location
        
    Returns:
        dict: JSON data from API or None if error occurs
    """
    
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "current_weather": True
    }
    
    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()  # Raise HTTPError for bad status
        data = response.json()
        pprint(data)  # Pretty-print raw JSON
        return data
    
    except requests.exceptions.HTTPError as http_err:
        print(f"HTTP error occurred: {http_err}")
        logging.error(f"HTTP error: {http_err}")
    except requests.exceptions.ConnectionError as conn_err:
        print(f"Connection error occurred: {conn_err}")
        logging.error(f"Connection error: {conn_err}")
    except requests.exceptions.Timeout as timeout_err:
        print(f"Timeout error occurred: {timeout_err}")
        logging.error(f"Timeout error: {timeout_err}")
    except requests.exceptions.RequestException as req_err:
        print(f"General request error occurred: {req_err}")
        logging.error(f"Request exception: {req_err}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        logging.error(f"Unexpected error: {e}")
    
    return None

if __name__ == "__main__":
    # Example: Kathmandu, Nepal
    latitude = 27.7
    longitude = 85.3
    
    # Fetch weather data
    weather_data = fetch_weather_data(latitude, longitude)
    
    if weather_data and "current_weather" in weather_data:
        # Flatten current weather JSON into DataFrame
        df = pd.json_normalize(weather_data["current_weather"])
        print("\nWeather DataFrame:\n")
        print(df)
    else:
        print("No weather data available.")
```
