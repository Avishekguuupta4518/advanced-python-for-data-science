```python
# Create a car class with attributes make, model, and year. Add a method to display car details.

class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_details(self):
        print(f"{self.year} {self.make} {self.model}")


# Object for Car
my_car = Car("Toyota", "Corolla", 2022)
my_car.display_details()
```

```python
#  Extend the Car class with an additional attribute battery_capacity. Add a method to display battery info.

class ElectricCar(Car):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def display_battery_info(self):
        print(f"Battery Capacity: {self.battery_capacity} kWh")


# Object for ElectricCar
my_electric = ElectricCar("Tesla", "Model S", 2023, 100)
my_electric.display_details()
my_electric.display_battery_info()
```

```python
# -  Create a class with attributes length and width. Add methods to calculate area and perimeter.

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Object for Rectangle
rect = Rectangle(10, 5)
print("Rectangle Area:", rect.area())
print("Rectangle Perimeter:", rect.perimeter())
```

```python
# - Create a class with attribute radius. Use encapsulation to protect radius and add getter and setter methods. Add a method to calculate the area

class Circle:
    def __init__(self, radius):
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def set_radius(self, radius):
        if radius > 0:
            self.__radius = radius
        else:
            print("Radius must be positive!")

    def area(self):
        import math
        return math.pi * self.__radius ** 2


# Object for Circle
circle = Circle(7)
print("Circle Radius:", circle.get_radius())
print("Circle Area:", round(circle.area(), 2))
circle.set_radius(10)
print("Updated Radius:", circle.get_radius())
print("Updated Area:", round(circle.area(), 2))
```

```python
# Create a class with methods for deposit, withdraw, and checking balance. Ensure the balance cannot be accessed directly.

class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}")
        else:
            print("Insufficient funds or invalid amount")

    def check_balance(self):
        return self.__balance


# Object for BankAccount
account = BankAccount()
account.deposit(500)
account.withdraw(200)
print("Current Balance:", account.check_balance())
```

```python
# - Create a class with attributes name, roll_number, and marks. Add methods to display student info and calculate average marks.

class Student:
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number
        self.marks = marks

    def display_info(self):
        print(f"Name: {self.name}, Roll Number: {self.roll_number}, Marks: {self.marks}")

    def average_marks(self):
        return sum(self.marks) / len(self.marks)


# Object for Student
student1 = Student("Avishek", "CS101", [85, 90, 78])
student1.display_info()
print("Average Marks:", round(student1.average_marks(), 2))
```

```python
# -  Implement a class that inherits from Person. Override the greet method to include the subject taught.

class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, I am {self.name}")


class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def greet(self):
        print(f"Hello, I am {self.name}, and I teach {self.subject}")


# Object for Teacher
teacher1 = Teacher("Mr. Sharma", "Mathematics")
teacher1.greet()
```

```python
# - Create a class with attributes title, author, and available. Add methods to borrow and return the book. Use encapsulation for available.

class LibraryBook:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.__available = True

    def borrow(self):
        if self.__available:
            self.__available = False
            print(f"You have borrowed '{self.title}'")
        else:
            print(f"'{self.title}' is currently not available")

    def return_book(self):
        self.__available = True
        print(f"'{self.title}' has been returned")

    def is_available(self):
        return self.__available


# Object for LibraryBook
book1 = LibraryBook("Python 101", "John Doe")
print("Available?", book1.is_available())
book1.borrow()
print("Available?", book1.is_available())
book1.return_book()
print("Available?", book1.is_available())
```

```python
# -  Create a Shape class and two subclasses Triangle and Square. Implement area calculation for both using inheritance.

class Shape:
    def area(self):
        pass


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


# Objects
triangle = Triangle(10, 5)
square = Square(6)
print("Triangle Area:", triangle.area())
print("Square Area:", square.area())
```

```python
# -  Implement a class with attributes name, salary, and position. Add methods to display info and give a raise. Demonstrate encapsulation for salary.

class CompanyEmployee:
    def __init__(self, name, salary, position):
        self.name = name
        self.__salary = salary
        self.position = position

    def display_info(self):
        print(f"Name: {self.name}, Position: {self.position}, Salary: ${self.__salary}")

    def give_raise(self, amount):
        if amount > 0:
            self.__salary += amount
            print(f"{self.name} received a raise of ${amount}")
        else:
            print("Raise must be positive")

    def get_salary(self):
        return self.__salary


# Object for CompanyEmployee
employee1 = CompanyEmployee("Avishek Gupta", 50000, "Software Engineer")
employee1.display_info()
employee1.give_raise(5000)
print("Updated Salary:", employee1.get_salary())
```
