```python
# Difference between append() and extend()
# append() → adds a single element
# extend() → adds elements from another iterable

lst = [1, 2]
lst.append([3, 4])  # [1, 2, [3, 4]]
lst.extend([5, 6])  # [1, 2, [3, 4], 5, 6]
```

```python
# *args and **kwargs
# *args → arbitrary positional arguments
# **kwargs → arbitrary keyword arguments

def demo(*args, **kwargs):
    print(args)
    print(kwargs)

demo(1, 2, 3, a=4, b=5)
```

```python
# Difference between try-except and if-else
# if-else → checks conditions beforehand
# try-except → handles runtime errors

x = "a"
# if-else won't catch runtime errors
if x != 0:
    print(10 / x)  # Error if x isn't number

# try-except catches errors
try:
    print(10 / x)
except TypeError:
    print("Cannot divide by non-number")
```
