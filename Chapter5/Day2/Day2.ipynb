```python
import schedule
import time
from datetime import datetime

def write_file():
    """Append the current timestamp to a daily log file."""
    try:
        # Create a separate log file for each day
        filename = datetime.now().strftime("schedule_%Y-%m-%d.txt")
        with open(filename, "a") as f:
            f.write(f"Hello! The time is {datetime.now()}\n")
        print(f"Logged time to {filename}")
    except Exception as e:
        print(f"Error writing file: {e}")

# Schedule the task every minute
schedule.every(1).minutes.do(write_file)

print("Scheduler started. Press Ctrl+C to stop.")

try:
    while True:
        schedule.run_pending()
        time.sleep(1)
except KeyboardInterrupt:
    print("Scheduler stopped.")
```
