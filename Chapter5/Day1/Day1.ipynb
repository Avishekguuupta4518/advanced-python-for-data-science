```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
X = np.arange(1,30,0.27)
y = 1.234*np.square(X) + 2.456

plt.figure(figsize=(6,4))
plt.plot(X,y, c='g')
plt.xlabel("X")
plt.ylabel("y")
plt.title("X vs y")
plt.grid()
plt.show()
```

```python
#  Plots using subplots

fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.plot(X, y, c='r')
ax1.set_xlabel("X")
ax1.set_ylabel("y")
ax1.set_title("X vs y");

ax2 = fig.add_subplot(2,2,2)
ax2.plot(X, y, c='g')
ax2.set_xlabel("X")
ax2.set_ylabel("y")
ax2.set_title("X vs y");

plt.tight_layout()
```

```python
fruits = ["Apple", "Banana", "Orange", "Guava"]
quantity = [100,120, 70, 130]
fig, axes = plt.subplots(2,2)
for i in range(2):
    for j in range(2):
        axes[i, j].bar(fruits, quantity, color=['Red', 'Yellow', 'Orange', 'Green'], alpha=0.7)
        axes[i, j].set_title(f"Plot {i}, {j}")
        axes[i, j].set_xticklabels(labels=fruits, rotation=45)
        
plt.tight_layout()
plt.legend()

# fig.subplots_adjust(wspace=0.5, hspace=0.5)
```

```python
import matplotlib.pyplot as plt 
import numpy as np 

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=20, color='skyblue', edgecolor='black')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Random Data")
plt.grid(True)
plt.legend()
plt.show()
```

```python
from sklearn.datasets import make_moons

X ,y = make_moons(
    n_samples=1000,
    noise=0.03,
    random_state=42
)

fig, axes = plt.subplots()
axes.scatter(X[:,0], X[:,1], c=y)
axes.text(x=0.7, y=1, s="Hello World", family='monospace', fontsize=20)
axes.set_title("Scatter Plot")
plt.savefig('scatter.pdf')
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True, as_frame=True)

fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].bar(
    x=['Setosa', 'Versicolor', 'Virginica'],
    height=y.value_counts(),
    color=['r', 'g', 'b']
)
axes[0, 0].set_title("Species Distribution", fontsize=8)

axes[0, 1].scatter(X['sepal length (cm)'], X['petal length (cm)'], c=y)
axes[0, 1].set_title("Sepal vs Petal Length", fontsize=8)

axes[1, 0].hist(X['sepal length (cm)'], bins=20, color="red")
axes[1, 0].set_title("Sepal Length Histogram", fontsize=8)

axes[1, 1].boxplot(X['petal length (cm)'], vert=False)
axes[1, 1].set_title("Petal Length Boxplot", fontsize=8)

plt.tight_layout()
plt.show()
```

```python
plt.rc("figure", figsize=(8,6))
plt.rc("font", family='monospace', weight='bold', size=8)
```
